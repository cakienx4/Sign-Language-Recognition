- Load data

```python
import joblib

X_train = joblib.load('../data/X_train.pkl')
X_test = joblib.load('../data/X_test.pkl')
y_train = joblib.load('../data/y_train.pkl')
y_test = joblib.load('../data/y_test.pkl')
```

- Load scale

```python
scaler = joblib.load('../models/scaler.pkl')

X_train_scale = scaler.transform(X_train)
X_test_scale = scaler.transform(X_test)
```

- Tạo model randomforest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier(random_state=42)

model_rf.fit(X_train_scale,y_train)
```

```python
acc_train = round(model_rf.score(X_train_scale, y_train)*100, 2)

print("Độ chính xác trên tập huấn luyện: ", acc_train)
```

```python
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay, precision_recall_fscore_support

y_pred = model_rf.predict(X_test_scale)
acc_test = round(accuracy_score(y_test, y_pred, normalize=True)*100, 2)

print('Độ chính xác trên tập kiểm thử: ', acc_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
```

```python
prec, rec, f1_score, _ = precision_recall_fscore_support(y_test, y_pred, average='macro', zero_division=0)

print('Precision: ', prec)
print('Recall: ', rec)
print('F1_score: ', f1_score)
```

```python
joblib.dump(model_rf, '../models/model_rf.pkl')
```
